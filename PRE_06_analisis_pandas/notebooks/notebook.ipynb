```python
import pandas as pd
import matplotlib.pyplot as plt
import os


INPUT_DRIVERS = "../data/drivers.csv"
INPUT_TIMESHEET = "../data/timesheet.csv"

OUTPUT_FOLDER = "../submission"


def main():

    os.makedirs(
        OUTPUT_FOLDER,
        exist_ok=True
    )


    # Leer archivos
    drivers = pd.read_csv(INPUT_DRIVERS)

    timesheet = pd.read_csv(INPUT_TIMESHEET)


    # unir información
    df = timesheet.merge(
        drivers,
        on="driverId",
        how="left"
    )


    # resumen por conductor
    summary = (
        df.groupby(
            [
                "driverId",
                "name",
                "certified",
                "wage-plan"
            ]
        )
        .agg(
            total_hours=("hours-logged","sum"),
            total_miles=("miles-logged","sum"),
            avg_hours=("hours-logged","mean"),
            avg_miles=("miles-logged","mean")
        )
        .reset_index()
    )


    # guardar csv
    summary.to_csv(
        f"{OUTPUT_FOLDER}/summary.csv",
        index=False
    )


    # top 10 conductores por millas

    top10 = (
        summary
        .sort_values(
            "total_miles",
            ascending=False
        )
        .head(10)
    )


    plt.figure(
        figsize=(10,6)
    )


    plt.bar(
        top10["name"],
        top10["total_miles"]
    )


    plt.xticks(
        rotation=45,
        ha="right"
    )


    plt.title(
        "Top 10 drivers por millas acumuladas"
    )


    plt.xlabel(
        "Driver"
    )


    plt.ylabel(
        "Miles"
    )


    plt.tight_layout()


    plt.savefig(
        f"{OUTPUT_FOLDER}/top10_drivers.png"
    )


    plt.close()



if __name__ == "__main__":
    main()
```
